```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.patches import Rectangle
```

```python
colors = mcolors.XKCD_COLORS
sort_colors = True

if sort_colors is True:
    names = sorted(
        colors, key=lambda c: tuple(mcolors.rgb_to_hsv(mcolors.to_rgb(c))))
else:
    names = list(colors)

replace_figure = False

f = open("README.md", "w")

cell1 = " view "
fix_cell1 = ""
i = 0
while i < 71:
    try:
        fix_cell1 = fix_cell1 + cell1[i]
    except:
        fix_cell1 = fix_cell1 + " "
    i += 1
cell2 = "name"
fix_cell2 = ""
i = 0
while i < 71:
    try:
        fix_cell2 = fix_cell2 + cell2[i]
    except:
        fix_cell2 = fix_cell2 + " "
    i += 1
f.write("|" + fix_cell1 + " | " + fix_cell2 + " | \n")

fix_cell1 = ""
i = 0
while i < 71:
    fix_cell1 = fix_cell1 + "-"
    i += 1
fix_cell2 = ""
i = 0
while i < 71:
    fix_cell2 = fix_cell2 + "-"
    i += 1
f.write("|" + fix_cell1 + " | " + fix_cell2 + " | \n")

for i, name in enumerate(names):
    file_name = name.split(' ')[0][5:]
    if len(name.split(' ')) > 1:
        for sub in name.split(' ')[1:]:
            file_name = file_name + '_' + sub
    file_name = file_name.replace("/", "-")

    if replace_figure:
        fig = plt.figure(figsize=(2, 0.3), dpi=30)
        ax = fig.add_subplot(111)
        rect1 = Rectangle((0, 0), 100, 25, color=name, edgecolor = None)
        ax.add_patch(rect1)
        ax.yaxis.set_visible(False)
        ax.xaxis.set_visible(False)
        ax.set_axis_off()
        plt.subplots_adjust(top = 1, bottom = 0, right = 1, left = 0, 
                    hspace = 0, wspace = 0)
        plt.margins(0,0)
        plt.savefig("view/"+file_name+".png", pad_inches = 0)
        plt.close()

    cell1 = "!["+name+"](view/"+file_name+".png)"
    fix_cell1 = ""
    i = 0
    while i < 71:
        try:
            fix_cell1 = fix_cell1 + cell1[i]
        except:
            fix_cell1 = fix_cell1 + " "
        i += 1
    cell2 = name
    fix_cell2 = ""
    i = 0
    while i < 71:
        try:
            fix_cell2 = fix_cell2 + cell2[i]
        except:
            fix_cell2 = fix_cell2 + " "
        i += 1
    f.write("|" + fix_cell1 + " | " + fix_cell2 + " | \n")
f.close()
```
