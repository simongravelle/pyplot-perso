```python
import math
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.patches import Rectangle
```

```python
colors = mcolors.XKCD_COLORS
sort_colors = True

if sort_colors is True:
    names = sorted(
        colors, key=lambda c: tuple(mcolors.rgb_to_hsv(mcolors.to_rgb(c))))
else:
    names = list(colors)

replace_figure = False

f = open("README.md", "w")
for i, name in enumerate(names):
    file_name = name.split(' ')[0][5:]
    if len(name.split(' ')) > 1:
        for sub in name.split(' ')[1:]:
            file_name = file_name + '_' + sub
    file_name = file_name.replace("/", "-")

    if replace_figure:
        fig = plt.figure(figsize=(2, 0.3), dpi=72)
        ax = fig.add_subplot(111)
        rect1 = Rectangle((0, 0), 100, 25, color=name, edgecolor = None)
        ax.add_patch(rect1)
        ax.yaxis.set_visible(False)
        ax.xaxis.set_visible(False)
        ax.set_axis_off()
        plt.xlim([0, 20])
        plt.ylim([0, 50])
        plt.subplots_adjust(top = 1, bottom = 0, right = 1, left = 0, 
                    hspace = 0, wspace = 0)
        plt.margins(0,0)
        plt.savefig("view/"+file_name+".png", pad_inches = 0)
        plt.close()

    f.write("| !["+name+"](view/"+file_name+".png) | "+name+" | \n")
f.close()
```
